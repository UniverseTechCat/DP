```python
import pandas as pd
```

```python
df = pd.read_csv('sample_1.csv', encoding = 'utf-8')
```

```python
df.head()
```

```python
df = df.dropna(subset=["접수일자"])
df["접수일자"] = df["접수일자"].astype(int)
df["접수일자"] = df["접수일자"].astype(int)
```

```python
df["접수일자"] = pd.to_datetime(df["접수일자"], format = "%Y%m%d", errors = 'raise')
```

```python
df["접수일자"]
```

```python
df['접수시각'] = pd.to_datetime(df['접수시각'], format='%H:%M:%S' , errors='coerce').dt.time
```

```python
df['접수시각'].head()
```

```python
def categorize (age):
    if 0 < age <= 10:
        return '0대'
    elif 10 < age <= 20:
        return '10대'
    elif 20 < age <= 30:
        return '20대'
    elif 30 < age <= 40:
        return '30대'
    elif 40 < age <= 50:
        return '40대'
    elif 50 < age <= 60:
        return '50대'
    elif 60 < age <= 70:
        return '60대'
    elif 70 < age <= 80:
        return '70대'
    elif 80 < age <= 81:
        return '80세'
    else:
        return '80세 이상'

df['검사나이'] = df['검사나이'].apply(categorize)
```

```python
df['검사나이']
```

```python
df['퇴실일자'] = df['퇴실일자'].astype('str')
df['퇴실일자'].dtype

df['퇴실일자'] = pd.to_datetime(df['퇴실일자'], format = '%Y%m%d', errors = 'coerce')
df['퇴실일자'] = df['퇴실일자'].dt.strftime("%Y-%m-%d")

df['퇴실시각'] = df['퇴실시각'].astype('str')
df['퇴실시각'].dtypes

df['퇴실시각'] = df['퇴실시각'].str.ljust(6, fillchar = '0')
df['퇴실시각'] = pd.to_datetime(df['퇴실시각'], format='%H%M%S', errors='coerce').dt.time
```

```python
df['입실시각'] = df['입실시각'].fillna(0).astype(float).astype(int).astype(str)
df['입실시각'] = df['입실시각'].str.zfill(4)
df['입실시각'] = df['입실시각'].astype(str)
df['입실시각'] = pd.to_datetime(df['입실시각'], format='%H%M', errors='coerce')



df['입실시각'] = df['입실시각'].dt.strftime('%H:%M:%S')


print(df['입실시각'].head())
```

```python
date1 = pd.to_datetime(df['입실일자'])
date2 = pd.to_datetime(df['퇴실일자'])

df['입원기간'] = (date2 - date1).dt.days.astype(str) + ' days'
```
