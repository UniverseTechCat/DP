```python
df =pd.DataFrame({
    "문항1":['1','2',"",'2'],
    "문항2":["",'2','3',""]
})
```

```python
df["문항"] = df["문항1"].fillna("")+","+df["문항2"].fillna("")
```

```python
df = df.drop(columns=["문항1","문항2"])
```

```python
print(df)
```

```python
df =pd.DataFrame({
    "문항1":['1','2',"",'2'],
    "문항2":["",'2','3',""],
    "문항3":["",'1','2',""]
})
```

```python
#공백 문자열 처리
cols = ["문항1","문항2","문항3"]
df[cols]=df[cols].replace(r'^\s*$','무응답',regex=True)
```

```python
#문항 합치기
df["문항"]=df["문항1"].fillna("")+","+df["문항2"].fillna("")+","+df["문항3"].fillna("")
```

```python
print(df)
```

```python
df = pd.DataFrame({
    'variable':['value1-value2-value3',"value3-value1-value5","vvdfvalue165656"]
})
```
